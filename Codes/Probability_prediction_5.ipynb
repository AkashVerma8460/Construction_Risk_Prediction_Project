```python
# ============================================================
# CODE 3: PROBABILITY PREDICTION
# ============================================================

import os
import joblib
import pandas as pd


# ------------------------------------------------------------
# 1. LOAD BEST MODELS
# ------------------------------------------------------------

delay_model = joblib.load(
    "models/best_delay_model.pkl"
)

cost_model = joblib.load(
    "models/best_cost_model.pkl"
)


# ------------------------------------------------------------
# 2. ENTER NEW PROJECT DATA
# ------------------------------------------------------------

# Change these values for each new building project.

project = {

    "Planned_Cost": 50000000,

    "Planned_Duration": 600,

    "Start_Year": 2026,

    "Start_Month": 8,

    "Location": "Kanpur",

    "Weather_Condition": "Sunny",

    "Temperature": 32,

    "Humidity": 65,

    "Air_Quality_Index": 150,

    "Material_Usage": 5000,

    "Labor_Hours": 25000,

    "Equipment_Utilization": 75,

    "Energy_Consumption": 12000,

    "Accident_Count": 2,

    "Safety_Risk_Score": 6,

    "Vibration_Level": 3,

    "Crack_Width": 0.2,

    "Load_Bearing_Capacity": 300
}


# ------------------------------------------------------------
# 3. CALCULATE START QUARTER
# ------------------------------------------------------------

project["Start_Quarter"] = (

    (project["Start_Month"] - 1) // 3 + 1

)


# ------------------------------------------------------------
# 4. CREATE DATAFRAME
# ------------------------------------------------------------

new_project = pd.DataFrame([project])


# ------------------------------------------------------------
# 5. PREDICT SCHEDULE DELAY
# ------------------------------------------------------------

delay_prediction = delay_model.predict(
    new_project
)[0]

delay_probability = delay_model.predict_proba(
    new_project
)[0, 1]


# ------------------------------------------------------------
# 6. PREDICT COST OVERRUN
# ------------------------------------------------------------

cost_prediction = cost_model.predict(
    new_project
)[0]

cost_probability = cost_model.predict_proba(
    new_project
)[0, 1]


# ------------------------------------------------------------
# 7. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("       BUILDING PROJECT RISK PREDICTION")
print("=" * 60)

print("\nSCHEDULE DELAY")

print(
    "Prediction:",
    "YES" if delay_prediction == 1 else "NO"
)

print(
    f"Probability of Delay: "
    f"{delay_probability * 100:.2f}%"
)


print("\nCOST OVERRUN")

print(
    "Prediction:",
    "YES" if cost_prediction == 1 else "NO"
)

print(
    f"Probability of Cost Overrun: "
    f"{cost_probability * 100:.2f}%"
)


# ------------------------------------------------------------
# 8. SAVE RESULT
# ------------------------------------------------------------

os.makedirs(
    "prediction_results",
    exist_ok=True
)


result = pd.DataFrame({

    "Schedule_Delay_Prediction": [
        "YES" if delay_prediction == 1 else "NO"
    ],

    "Schedule_Delay_Probability": [
        delay_probability * 100
    ],

    "Cost_Overrun_Prediction": [
        "YES" if cost_prediction == 1 else "NO"
    ],

    "Cost_Overrun_Probability": [
        cost_probability * 100
    ]

})


result.to_csv(
    "prediction_results/project_prediction.csv",
    index=False
)


print(
    "\nResult saved to:"
    " prediction_results/project_prediction.csv"
)
```
